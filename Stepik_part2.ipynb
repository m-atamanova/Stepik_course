```python
#На вход программе подаётся последовательность слов, каждое слово на отдельной строке.
#Концом последовательности является слово «КОНЕЦ» (без кавычек).
#При этом само слово «КОНЕЦ» не входит в последовательность, лишь символизируя её окончание.
#Напишите программу, которая выводит члены данной последовательности.
text = input()
while text != 'КОНЕЦ':
    print(text)
    text = input()
```

```python
#На вход программе подаётся последовательность слов, каждое слово на отдельной строке.
#Концом последовательности является слово «КОНЕЦ» или «конец» (без кавычек).
#При этом сами слова «КОНЕЦ» и «конец» не входят в последовательность, лишь символизируя её окончание.
#Напишите программу, которая выводит члены данной последовательности.
text = input()
while (text != 'КОНЕЦ') and (text != 'конец'):
    print(text)
    text = input()
```

```python
#На вход программе подаётся последовательность слов, каждое слово на отдельной строке.
#Концом последовательности является одно из трёх слов: «стоп», «хватит», «достаточно» (без кавычек). Сами эти слова в последовательность не входят, лишь символизируя её окончание.
#Напишите программу, которая выводит общее количество членов данной последовательности.
text = input()
i = 0
while text != 'стоп' and text != 'хватит' and text != 'достаточно':
    i += 1
    text = input()
print(i)
```

```python
#На вход программе подаётся последовательность целых чисел делящихся на 7,
#каждое число на отдельной строке. Концом последовательности является любое число, не делящееся на 7
#(само это число в последовательность не входит, лишь символизируя её конец).
#Напишите программу, которая выводит члены данной последовательности.
n = int(input())
while n % 7 == 0:
    print(n)
    n = int(input())
```

```python
#На вход программе подаётся последовательность целых чисел, каждое число на отдельной строке.
#Признаком окончания последовательности является любое отрицательное число, при этом в саму последовательность оно не входит.
#Напишите программу, которая выводит сумму всех членов данной последовательности.
n = int(input())
total_sum = 0
while n >= 0:
    total_sum += n
    n = int(input())
print(total_sum)
```

```python
#На вход программе подаётся последовательность целых чисел от 1 до 5,
#характеризующее оценку ученика, каждое число на отдельной строке.
#Концом последовательности является любое неположительное число либо число, большее 5.
#Напишите программу, которая выводит количество пятерок.
n = int(input())
count = 0
while n > 0 and n <= 5:
    if n == 5:
        count += 1
    n = int(input())
print(count)
```

```python
#Всем известно, что ведьмак способен одолеть любых чудовищ, однако его услуги обойдутся недешево.
#К тому же ведьмак не принимает купюры, он принимает только чеканные монеты.
#В мире ведьмака существуют монеты с номиналами 1,5,10,25
#Напишите программу, которая определяет, какое минимальное количество чеканных монет нужно заплатить ведьмаку.
price = int(input())
count5 = 0
count10 = 0
count25 = 0
while price >= 25:
    count25 += 1
    price = price - 25
while price >= 10:
    count10 += 1
    price = price - 10
while price >= 5:
    count5 += 1
    price = price -5
total = price + count5 + count10 +count25
print(total)
```

```python
#Дано натуральное число n(n≥10).
#Напишите программу, которая определяет его максимальную и минимальную цифры и выводит текст в следующем формате:
n = int(input())
Max, Min = 0, 9
while n != 0:
    last_digit = n % 10
    if last_digit > Max:
        Max = last_digit
    n = n // 10
    if last_digit < Min:
        Min = last_digit
print(f'Максимальная цифра равна {Max}')
print(f'Минимальная цифра равна {Min}')
```

```python
#На вход программе подаются два натуральных числа a и b (a< b).
#Напишите программу, которая находит натуральное число из отрезка [a;b] (от a до b включительно)
#с максимальной суммой делителей.
#Если чисел с максимальной суммой делителей несколько, то искомым числом является наибольшее из них.
msum = 0                              # Максимальная сумма
sum = 0                               # Сумма делителей числа
x = 0                                 # Число с максимальной суммой делителей
a, b = int(input()), int(input())
for i in range(a, b+1):               # На отрезке от a до b
    for j in range(1, i+1):           # Считаем сумму делителей числа (от 1 до этого числа)
        if i % j == 0:
            sum += j
            if sum >= msum:
                msum = sum
                x = j                 # Сохраняем число с наибольшей суммой делителей
    sum = 0
print(x, msum)
```

```python
#написать программу, которая поможет Сэму проверять комментарии.
#Программа должна принимать на вход натуральное число n, а затем n строк, представляющих тексты комментариев.
#Для каждого комментария ваша программа должна выводить
#-номер этого комментария (начиная с 1),
#-затем двоеточие (:),
#-затем через пробел его текст или сообщение «COMMENT SHOULD BE DELETED» (без кавычек),
#если комментарий должен быть удалён Сэмом.
n = int(input())
for i in range(1, n+1):
    s = input()
    if s.isspace() == True or s == '':
        print(f'{i}: COMMENT SHOULD BE DELETED')
    else:
        print(f'{i}: {s}')
```

```python
#Напишите программу, которая выводит «Correct»,
#если никнейм соответствует всем вышеприведенным правилам, или «Incorrect» в противном случае.
#-никнейм должен начинаться с символа @
#-никнейм должен содержать от 5 до 15 (включительно) символов (включая первый символ @)
#-никнейм должен содержать только строчные буквы и (или) цифры (помимо первого символа @)
s = input()
Flag = True
if 5 <= len(s) <= 15:
    if s[0] == '@':
        s = s[1::]
    else:
        Flag = False
    if s.isalnum() != True:
        Flag = False
    if s.lower() != s:
        Flag = False
else:
    Flag = False

if Flag == True:
    print('Correct')
else:
    print('Incorrect')
```

```python
#Напишите программу, которая принимает на вход строку и проверяет,
#является ли эта строка корректным автомобильным номером.
s = input()
Flag = True

if len(s) != 9 and len(s) != 10:
    Flag = False
else:
    letter = s[0] + s[4] + s[5]
    digit = s[1] + s[2] + s[3] + s[7:]

    for i in range(3):
        if letter[i] not in 'АВЕКМНОРСТУХ':
            Flag = False
            break
    for i in range(len(digit)):
        if digit[i] not in '0123456789':
            Flag = False
            break
    if s[6] != '_':
        Flag = False

if Flag == False:
    print('NO')
else:
    print('YES')
```

```python
#На вход программе подаётся строка текста.
#Требуется написать программу, которая находит стоимость старого и нового сообщений Сэма в 🐝
#(в новом - англ.буквы меняются на русские)
total_old = 0
s = input()
for i in range(len(s)):
    total_old += ord(s[i]) * 3
eng = "eyopaxcETOPAHXCBM"
rus = "еуорахсЕТОРАНХСВМ"
a = s
for j in range(len(eng)):
    a = a.replace(eng[j], rus[j])
total_new = 0
for k in range(len(a)):
    total_new += ord(a[k]) * 3
print(f'Старая стоимость: {total_old}🐝')
print(f'Новая стоимость: {total_new}🐝')
```

```python
#На вход программе подаётся строка текста.
#Расшифруйте текст, заменив все конструкции [u-<номер символа в таблице Unicode>] на соответствующие буквы русского алфавита.
s = input()
c = s
for i in range(len(s)-3):
    if s[i] == '[' and s[i+1] == 'u' and s[i+2] == '-':
        k = i+3
        a = ''
        while s[k] != ']':
            a += s[k]
            k += 1
        b = chr(int(a))

        c = c.replace(a, b)
        c = c.replace('[u-', '')
        c = c.replace(']', '')

print(c)
```

```python
#Шифр Цезаря
n = int(input())
s = input()
a = ''
for i in range(len(s)):
    b = ord(s[i])
    e = b-n
    if e < 97:
        d = b - 97
        m = n - d
        c = chr(123 - m)
        a += c
    else:
        a += chr(e)
print(a)
```

```python
#На вход программе подаётся последовательность строк, каждая строка на отдельной строке.
#Концом последовательности является слово «КОНЕЦ» (без кавычек).
#Напишите программу, которая находит в данной последовательности максимальную и минимальную строки
M = ''
s = input()
m = s
while s != 'КОНЕЦ':
    if s >= M:
        M = s
    if s <= m:
        m = s
    s = input()
print(f'Минимальная строка ⬇️: {m}')
print(f'Максимальная строка ⬆️: {M}')
```
